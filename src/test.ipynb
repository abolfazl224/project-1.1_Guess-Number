```python
import random
```

```python
random_number = random.randint(1, 100)
print(random_number)
```

```python
class score:
    def __init__(self, score):
        self.score = score

    def decrease_score(self):
        self.score = self.score - 10

    def result(self):
        return f'the score is {self.score}'        
```

```python
def valid_input(start, end):
    while True:
        user_input = input("please enter a number(for ending enter 'q'):")
        if not user_input.isdigit():
            if user_input == 'q':
                print('Thank you for choose my game. Good By.')
                break
    
            else:
                print('please enter valid input')
                continue

        user_input = int(user_input)
        if (user_input < start) or (user_input > end):
            print(f'please enter a number between {start} and {end}')
            continue
        
        return user_input
```

```python
my_score = score(100)
while True:
    user_number = valid_input(1, 100)
    if not user_number:
        break
    
    if hint(user_number, random_number):
        print(f"your score is {my_score.result()}")
        break

    else:
        my_score.decrease_score()    
```

```python
def hint(user:int, system:int):
    if user == system:
        print(f'your guess is correct')
        return True
        
    elif user > system :
        print('Please decrease your guess')
        return False

    else :
        print('Please increase your guess')
        return False 
```
